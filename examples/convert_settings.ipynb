# Converting manually created configuration files

```python
# Make sure you have the most recent version installed.
#%pip install -e ..
```

```python
from biopap.tools.update_config import ConfigurationUpdater
```

When you manually create a configuration file, it will look something like this:

For BioPAP to be able to automatically create and run a pipeline from this configuration, each step needs a full reference to the relevant BioPAP module. To add those references, you can call the ConfigurationUpdater. This tool will resolve and add the relevant module from each of the configuration keys - Anonymize, Rereference, Plot_1 etc.

**Please note**: as this process requires that the keys can still be recognized as the module they represent, use only postfixes to differentiate between instances of the same module. For instance, `Plot_2` and `Plot_Final` are okay for different instances of `Plot`, but `Final_Plot` is not. If the ConfigurationUpdater encounters a key that cannot be resolved, it will raise an error.

You can run the ConfigurationUpdater on your configuration file, or directly on its contents:

```python
config = """
EOGChannels:
    ref_channels: ['hEOG', 'vEOG']
    anode_channels: ['EXG3', 'EXG5']
    cathode_channels: ['EXG4', 'EXG6']

IIRFilter:
    l_freq: 0.1
    h_freq: 35
    
Plot_1:
  plot_types: 31
  save_folder: /data/BioPAP/out/out
  plot_sensors:
    kind: '3d'
  plot_topomap:
    image_interp: 'nearest'
"""
```

```python
new_config = ConfigurationUpdater().convert(config)
```

```python
# To use the update configuration, copy/paste the output to your new configuration file.
print(new_config)
```

```python
# You can also convert a configuration file:
ConfigurationUpdater().convert_file('/path/to/config.yaml')
# which will write a new file: /path/to/config_new.yaml'
```
